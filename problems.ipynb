## Problem 1: Representing SHA-256 into python code.
SHA-256 is specified in [NIST FIPS 180-4](https://csrc.nist.gov/files/pubs/fips/180-4/final/docs/fips180-4.pdf), which describes how messages shorter than 2<sup>64</sup> bits are processed to produce a 256-bit message digest, which is the hash output. 

### Representing 32-bit words
SHA-256 uses fixed-width 32-bit words, whereas Python's built-in `int` type can represent integers of arbitrary size. The [Python documentation](https://docs.python.org/3/library/stdtypes.html#bitwise-operations-on-integer-types) explains that the bitwise AND operator `&` can be used with the mask `0xFFFFFFFF` to retain the lowest 32 bits of an integer. This section examines how Python data types can represent the input, output, and intermediate data structures used in SHA-256, without implementing the hashing operations or message padding.
